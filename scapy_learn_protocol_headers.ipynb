## RFC

RFC stands for "Request for Comments". They are formal documents that track standards development published by the Internet Engineering Task Force (IETF). Information on these RFCs can be found at https://www.ietf.org/process/rfcs/

In Scapy, view an "RFC-like representation" with **rfc( )**

## Scapy commands used in this notebook

Use the following import statement to be able to call Scapy functions without having to always write "**scapy.**"

```python
from scapy.all import *
```

### _lsc( )_

- The lsc( ) function helps when you can't quite remember the name of the command you want to use
- The name of the command is left of the colon
- A brief description of the command is on the right

```python
lsc() # to list all Scapy commands
```

### _ls( )_

- With no argument, ls( ) lists all protocols
- On the left is the variable name of the protocol that can then be used as an argument
- This is useful when you are unsure how a protocol is named in Scapy

```python
ls() # to list all protocols
```

### _rfc( protocol)_

- To view an RFC-like representation of protocol fields

```python
rfc(IP) # RFC-like representation, ex. for IPv4
```

### _ls(protocol)_

- **ls(protocol):** When ls( ) is used with a protocol as an argument, you can view the Scapy default values for the protocol fields
- Notice that the argument is not in quotes
- When in doubt, to find the Scapy variable name for the protocol, first just use ls( ) with no argument, then use ls(protocol) with the protocol you determined to be the one you needed

```python
ls(IP) # to view the Scapy default field values, ex. for IPv4
```

## Data Link Layer

### _Address Resolution Protocol (ARP)_

```python
rfc(ARP) # rfc representation for ARP
```

The Scapy default values include:
- "hwtype": Hardware type. 1 means the hearder is in Ethernet format. There are 33 hardware type options for ARP

- "ptype": The intended layer 3 protocol type. 2048, the decimal value of 0x800, is the value for IPv4

- "op": Operation, or the objective. 1 means request, 2 means response

```python
ls(ARP) # Scapy ARP default values
```

**ARP header info:** 
- https://en.wikipedia.org/wiki/Address_Resolution_Protocol#Packet_structure
- https://www.frameip.com/entete-arp/#3-8211-definition-des-differents-champs

### _Ethernet_

```python
rfc(Ether) # rfc representation for Ethernet
```

**The Scapy default value:**
- "type": (Ether)type. 36864 is the decimal version of 0x9000, the value for Ethernet Configuration Testing Protocol

```python
ls(Ether) # Scapy Ethernet default values
```

**Ethernet II header info:**
- https://en.wikipedia.org/wiki/Ethernet_frame#Ethernet_II
- https://en.wikipedia.org/wiki/EtherType
- https://www.frameip.com/entete-ethernet/#4-8211structure-de-lrsquoentete-ethernet

## Network Layer

### _Internet Control Message Protocol (ICMP)_

```python
rfc(ICMP) # rfc representation for ICMP
```

**The Scapy default values include:**
- "type": Type of control message. 8 means echo request
- "code": Code. Sub-code 0 means ping echo request

```python
ls(ICMP) # Scapy default ICMP values
```

**ICMP header info:**
- https://en.wikipedia.org/wiki/Internet_Control_Message_Protocol#Header
- https://www.frameip.com/entete-icmp/#2-8211-structure-de-lrsquoentete-icmp
- https://fr.wikipedia.org/wiki/Internet_Control_Message_Protocol#Format_des_paquets

### _Internet Protocol version 4 (IPv4)_

```python
rfc(IP) # rfc representation for IPv4
```

**The Scapy default values include:**
- "version": Version. Set at 4 to indicate IPv4.
- "ttl": Time to Live. 64 means the maximum number of hops. At each router, the ttl is decremented by one until it reaches zero, whereby the packet is discarded and a time-is-exceeded ICMP is sent.

```python
ls(IP) # Scapy default IPv4 values
```

**IPv4 header info:** 
- https://en.wikipedia.org/wiki/IPv4#Header
- https://www.frameip.com/entete-ip/#2-8211-structure-de-lrsquoentete
- https://fr.wikipedia.org/wiki/IPv4#En-t%C3%AAte_IPv4

### _Internet Protocol Version 6 (IPv6)_

```python
rfc(IPv6) # rfc representation for IPv6
```

**The Scapy default values include:**
- "version": Version. Set at 6 to indicate IPv6.
- "nh": Next Header. 59 indicates "No next header for IPv6"
- "hlim": Hop Limit. Replaces Time to Live. 64 means the maximum number of hops. At each "forwarding node", the hlim is decremented by one until it reaches zero.

```python
ls(IPv6) # Scapy default IPv6 values
```

**IPv6 header info:** 
- https://en.wikipedia.org/wiki/IPv6_packet#Fixed_header
- https://en.wikipedia.org/wiki/List_of_IP_protocol_numbers
- https://www.frameip.com/entete-ipv6/#2-8211structure-de-lrsquoentete-ipv6

## Application Layer

On Stack Overflow, there is a response that explains application layer protocols like HTTP, FTPand Telnet use the TCP format:
https://stackoverflow.com/questions/67404757/recognize-telnet-protocol-with-scapy-python

The way Scapy determines the application protocol running on TCP would be via the port used. Examples of known ports are:

21 => File Transfer Protocol (FTP)

22 => Secure Socket Shell (SSH)

23 => Telnet

80 => HyperText Transfer Protocol (HTTP)

If a user is trying to disguise a SSH session by using another port number, a deeper analysis of the capture would be needed to determine if it is indeed a SSH packet.

**For info on ports 21, 22, 23, etc., and other known ports numbers:**
- https://en.wikipedia.org/wiki/List_of_TCP_and_UDP_port_numbers#Well-known_ports
